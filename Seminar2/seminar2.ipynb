```python
penguins_numerical=penguins[penguins.columns[2:7]].groupby("sex")
```

```python
penguins_numerical.aggregate("mean") #agregare dupa medie
```

```python
male_penguins=penguins_numerical.get_group("Male")
female_penguins=penguins_numerical.get_group("Female")
```

```python
male_penguins.head()
```

```python
numeric_male_pengiun=male_penguins.select_dtypes(include="number") #sau exclude="str"
numeric_male_pengiun.fillna(numeric_male_pengiun.mean())

numeric_female_pengiun=female_penguins.select_dtypes(include="number") #sau exclude="str"
numeric_female_pengiun.fillna(numeric_female_pengiun.mean())

```
