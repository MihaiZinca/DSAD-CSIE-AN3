6. Să se calculeze, pentru fiecare județ și fiecare activitate, indicele de localizare și să se salveze rezultatul în fișierul *Cerinta6.csv*.

   Indicele de localizare arată cât de concentrată este o activitate într-un județ, raportat la mărimea județului. Se calculează ca raport între:
   * ponderea județului *i* în numărul total de angajați din activitatea *j* la nivel național, și
   * ponderea județului *i* în populația totală a țării.

   $$IL_{i,j} = \frac{A_{i,j} / A_j}{P_i / P}$$

   unde $A_{i,j}$ este numărul de angajați din activitatea $j$ în județul $i$, $A_j$ este numărul total de angajați din activitatea $j$ la nivel național, $P_i$ este populația județului $i$, iar $P$ este populația totală.

   O valoare $IL > 1$ înseamnă că activitatea este mai concentrată în județ decât ar justifica populația acestuia.

   **Exemplu:** județul *x* are 3.000 de angajați în industria cărnii și 200.000 de locuitori. La nivel național sunt 100.000 de angajați în industria cărnii și 20.000.000 de locuitori.

   $$IL_{x,\text{Carne}} = \frac{3\,000 / 100\,000}{200\,000 / 20\,000\,000} = \frac{0{,}03}{0{,}01} = 3$$

   Așadar, județul *x* are de 3 ori mai mulți angajați în industria cărnii decât ar fi de așteptat după populația sa.
